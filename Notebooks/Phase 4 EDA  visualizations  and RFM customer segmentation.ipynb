# Phase 02 - Data Cleaning

### Remove Duplicate Rows

```python
# Remove exact duplicate rows, keep the first occurrence
print("Before removing duplicates:", df.shape)

df = df.drop_duplicates()

print("After removing duplicates:", df.shape)
```

### Handle Missing Customer ID

```python
# Customer ID is essential for customer-level analysis (like RFM segmentation later)
# Rows without a Customer ID can't be linked to a specific customer, so we remove them
print("Before removing missing Customer ID:", df.shape)

df = df.dropna(subset=["Customer ID"])

print("After removing missing Customer ID:", df.shape)
```

### Handle Missing Description

```python
# A few rows have missing product descriptions
# Since it's a small number, we can safely drop these rows too
print("Before removing missing Description:", df.shape)

df = df.dropna(subset=["Description"])

print("After removing missing Description:", df.shape)
```

### Separate Cancelled Orders (Negative Quantity)

```python
# Invoices starting with "C" are cancellations/returns (this is how the dataset marks them)
# We save these separately in case we want to analyze returns later,
# then keep only valid sales transactions in the main dataframe

df["Invoice"] = df["Invoice"].astype(str)

cancelled_orders = df[df["Invoice"].str.startswith("C")]
print("Cancelled/returned orders:", cancelled_orders.shape[0])

df = df[~df["Invoice"].str.startswith("C")]
print("Remaining valid sales orders:", df.shape[0])
```

### Remove Invalid Prices

```python
# Price should never be 0 or negative for a real sale
print("Before removing invalid prices:", df.shape)

df = df[df["Price"] > 0]

print("After removing invalid prices:", df.shape)
```

### Fix Data Types

```python
# Convert InvoiceDate from text to proper datetime format
# This lets us do date-based analysis later (monthly trends, etc.)
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

# Confirm the change
df.info()
```

### Create a Useful New Column

```python
# TotalPrice = how much revenue this line item generated
# This will be very useful for sales analysis in the next phase
df["TotalPrice"] = df["Quantity"] * df["Price"]

df.head()
```

### Final Check & Save Cleaned Data

```python
# Confirm everything is clean now
print("Final shape:", df.shape)
print("\nRemaining missing values:\n", df.isnull().sum())
print("\nRemaining duplicates:", df.duplicated().sum())
print("\nMinimum Quantity:", df["Quantity"].min())
print("Minimum Price:", df["Price"].min())
```

# Phase 3 (SQL)

### Load Cleaned Data & Create SQL Database

```python
import sqlite3
import os

# Load the cleaned data we saved in Phase 2
cleaned_path = "/content/drive/MyDrive/Retail Sales & Customer Insights Project/Dataset/cleaned_retail_data.csv"
df = pd.read_csv(cleaned_path)

# Create the Database folder if it doesn't already exist
db_folder = "/content/drive/MyDrive/Retail Sales & Customer Insights Project/Database"
os.makedirs(db_folder, exist_ok=True)

conn = sqlite3.connect(f"{db_folder}/retail_data.db")
df.to_sql("sales", conn, if_exists="replace", index=False)

print("Data loaded into SQL table 'sales'")
```

### Basic Overview Query

```python
# Total revenue and total number of orders
query = """
SELECT
    COUNT(DISTINCT Invoice) AS total_orders,
    ROUND(SUM(TotalPrice), 2) AS total_revenue
FROM sales
"""
pd.read_sql(query, conn)
```

### Top 10 Selling Products

```python
# Which products sold the most by quantity
query = """
SELECT
    Description,
    SUM(Quantity) AS total_quantity_sold,
    ROUND(SUM(TotalPrice), 2) AS total_revenue
FROM sales
GROUP BY Description
ORDER BY total_quantity_sold DESC
LIMIT 10
"""
pd.read_sql(query, conn)
```

### Monthly Revenue Trend

```python
# Revenue by month - useful to spot seasonality
query = """
SELECT
    strftime('%Y-%m', InvoiceDate) AS month,
    ROUND(SUM(TotalPrice), 2) AS monthly_revenue
FROM sales
GROUP BY month
ORDER BY month
"""
pd.read_sql(query, conn)
```

### Top 10 Customers by Spend

```python
# Which customers generated the most revenue
query = """
SELECT
    "Customer ID",
    ROUND(SUM(TotalPrice), 2) AS total_spent,
    COUNT(DISTINCT Invoice) AS total_orders
FROM sales
GROUP BY "Customer ID"
ORDER BY total_spent DESC
LIMIT 10
"""
pd.read_sql(query, conn)
```

### Sales by Country

```python
# Revenue breakdown by country
query = """
SELECT
    Country,
    ROUND(SUM(TotalPrice), 2) AS total_revenue,
    COUNT(DISTINCT Invoice) AS total_orders
FROM sales
GROUP BY Country
ORDER BY total_revenue DESC
"""
pd.read_sql(query, conn)
```

### Repeat Customer Rate

```python
# How many customers bought more than once vs. only once
query = """
SELECT
    CASE WHEN order_count > 1 THEN 'Repeat Customer' ELSE 'One-Time Customer' END AS customer_type,
    COUNT(*) AS number_of_customers
FROM (
    SELECT "Customer ID", COUNT(DISTINCT Invoice) AS order_count
    FROM sales
    GROUP BY "Customer ID"
)
GROUP BY customer_type
"""
pd.read_sql(query, conn)
```

## RFM Analysis (Customer Segmentation)

### Calculate RFM Metrics

```python
# RFM = Recency, Frequency, Monetary
# Recency: how many days since the customer's last purchase
# Frequency: how many orders the customer has placed
# Monetary: how much money the customer has spent in total

# Set a reference date (one day after the last transaction in the dataset)
reference_date = pd.to_datetime(df["InvoiceDate"]).max() + pd.Timedelta(days=1)

df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

rfm = df.groupby("Customer ID").agg(
    Recency=("InvoiceDate", lambda x: (reference_date - x.max()).days),
    Frequency=("Invoice", "nunique"),
    Monetary=("TotalPrice", "sum")
).reset_index()

rfm.head()
```

### Score Customers (1-4 scale)

```python
# Split customers into 4 groups (quartiles) for each metric
# For Recency, lower is better (recent buyers get higher score)
# For Frequency and Monetary, higher is better

rfm["R_Score"] = pd.qcut(rfm["Recency"], 4, labels=[4, 3, 2, 1])
rfm["F_Score"] = pd.qcut(rfm["Frequency"].rank(method="first"), 4, labels=[1, 2, 3, 4])
rfm["M_Score"] = pd.qcut(rfm["Monetary"], 4, labels=[1, 2, 3, 4])

rfm["RFM_Score"] = rfm["R_Score"].astype(str) + rfm["F_Score"].astype(str) + rfm["M_Score"].astype(str)

rfm.head()
```

### Label Customer Segments

```python
# Turn the RFM scores into readable business labels
def segment_customer(row):
    if row["R_Score"] == 4 and row["F_Score"] == 4 and row["M_Score"] == 4:
        return "Best Customers"
    elif row["F_Score"] == 4:
        return "Loyal Customers"
    elif row["R_Score"] == 4:
        return "Recent Customers"
    elif row["R_Score"] == 1:
        return "At Risk / Lost"
    else:
        return "Regular Customers"

rfm["Segment"] = rfm.apply(segment_customer, axis=1)

rfm["Segment"].value_counts()
```

### Visualize Customer Segments

```python
segment_counts = rfm["Segment"].value_counts()

plt.bar(segment_counts.index, segment_counts.values, color="teal")
plt.xticks(rotation=30, ha="right")
plt.title("Customer Segments (RFM Analysis)")
plt.xlabel("Segment")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```
